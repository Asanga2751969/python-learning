```python
import math
force_n = 12_000
diameter_mm = 10.0

area_mm2 = math.pi * diameter_mm**2/4
stress_mpa = force_n / area_mm2
print(f"Area: {area_mm2:.2f} mm2")
print(f"Stress: {stress_mpa:.1f} MPa")
```

```python
Pressure_PSI = 150
mpa_per_psi = 0.00689476
bar_per_mpa = 10
Pressure_MPa = Pressure_PSI * mpa_per_psi
Pressure_Bar = Pressure_MPa * bar_per_mpa
print(f"{Pressure_PSI} psi = {Pressure_MPa:.2f} MPa =  {Pressure_Bar:.2f} Bar")
```

```python
import math  #importing the math module provides access to the functions such as pi
total_load_kN = 8.5 #total shear load on the two bolts
bolt_diameter_mm = 8 #bolt diameter in milli meters
allowable_shear_stres_MPa = 300 #shear stress allowable in MPa
load_per_bolt_N = total_load_kN *1000 / 2 #multiplied by 1000 to converto KN to Newtons and divided by 2 because total load is divided between the two bolts
area_mm2 = math.pi * bolt_diameter_mm**2/4 # here the ** is the power operator  so bolt_diameter_mm**2 means the bolt_diameter_mm variable was squared
shear_stress_MPa = (load_per_bolt_N / area_mm2) #load per bolt divided by cross sectinal area of bolt
margin_of_safety = (allowable_shear_stres_MPa / shear_stress_MPa) - 1  #margin of safety

print(f"Load per Bolt in Newton = {load_per_bolt_N: .0f} N") #display load per bolt rounded off
print(f"Shear area single bolt = {area_mm2:.2f} mm2") #display single cross sectional area of bolt to 2 decimal places (.2f)
print(f"Shear Stress in MPa = {shear_stress_MPa:.2f} MPa") #display shear stress rounded off to 2 decimal places (.2f)
print(f"Margin of Safety = {margin_of_safety:.2f}") #display margin of safety rounded off to 2 decimal places (.2f)
```
