```python
margin = 0
if margin > 0 : #the question; note the colon at the end
  print ("Pass, margin is zero or positive") #indented, runs only if the answer is True
else:            #note the colon here too
  print ("Fail, margin is negative") #indented, runs only if the answer is False
```

```python
margin = -0.2
if margin > 0: # runs if the first question is true
  print("Pass, margin is positive")
elif margin < 0: # runs only if first condition was False and this one is True
  print("Fail, margin is negative")
elif margin == 0: # runs only if first two conditions were False and this one is True
  print("Pass, margin is zero")
else:
  print("What the hell is this crap?")
```

```python
for margin in [2.55, 0, -0.2, float("nan")]:
  if margin > 0: # runs if the first question is true
   print(f"{margin} Pass, margin is positive")
  elif margin < 0: # runs only if first condition was False and this one is True
    print(f"{margin} Fail, margin is negative")
  elif margin == 0: # runs only if first two conditions were False and this one is True
    print(f"{margin} Pass, margin is zero")
  else:
   print(f"{margin} ,Invalid number")

print("done")
```
